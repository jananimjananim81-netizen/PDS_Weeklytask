```python
# Numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Categorical columns
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
df[df.duplicated()]
```

```python
# Remove duplicates
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
categorical_columns = [
    "gender",
    "parental level of education",
    "lunch",
    "test preparation course"
]

for col in categorical_columns:
    print("\n", col)
    print(df[col].value_counts())
```

```python
for col in categorical_columns:
    print(col, ":", df[col].unique())
```

```python
for col in categorical_columns:
    df[col] = df[col].str.strip()
```

```python
df.dtypes
```

```python
score_columns = [
    "math score",
    "reading score",
    "writing score"
]
```

```python
mean_scores = df[score_columns].mean()

print("Mean Scoress:")
print(mean_scores)
```

```python
median_scores = df[score_columns].median()

print("Median Scores:")
print(median_scores)
```

```python
std_scores = df[score_columns].std()

print("Standard Deviation:")
print(std_scores)
```

```python
quartiles = df[score_columns].quantile([0.25, 0.50, 0.75])
quartiles
```

```python
score_statistics = df[score_columns].describe().T

score_statistics
```

```python
df["Total Score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
)

df[["math score", "reading score", "writing score", "Total Score"]].head()
```

```python
# Create Average Score
df["Average Score"] = df[score_columns].mean(axis=1)

df[["math score", "reading score", "writing score", "Average Score"]].head()
```

```python
# Create Percentage Performance
df["Percentage Performance"] = (
    df["Total Score"] / 300
) * 100

df["Percentage Performance"] = df["Percentage Performance"].round(2)
df[["Total Score", "Average Score", "Percentage Performance"]].head()
```

```python
pass_mark = 40

df["Pass/Fail Status"] = np.where(
    (df["math score"] >= pass_mark) &
    (df["reading score"] >= pass_mark) &
    (df["writing score"] >= pass_mark),
    "Pass",
    "Fail"
)
df["Pass/Fail Status"].value_counts()
```

```python
# Percentage of students in each category:
pass_percentage = (
    df["Pass/Fail Status"]
    .value_counts(normalize=True) * 100
)

pass_percentage
```

```python
# correlation
correlation_matrix = df[score_columns].corr()

correlation_matrix
```

```python
gender_performance = df.groupby("gender", observed=True)[
   score_columns + ["Average Score"]].mean()

gender_performance
```

```python
prep_performance = df.groupby(
    "test preparation course",
    observed=True
)["Average Score"].mean()

prep_performance
```

```python
lunch_performance = df.groupby(
    "lunch",
    observed=True
)["Average Score"].mean()

lunch_performance
```

```python
parent_education_performance = df.groupby(
    "parental level of education",
    observed=True
)["Average Score"].mean().sort_values(ascending=False)

parent_education_performance
```
